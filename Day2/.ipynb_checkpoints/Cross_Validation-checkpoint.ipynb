# Day 2, Week 4 Hands-On Lab: Cross-Validating a Model

## Overview & Learning Objectives
In this lab, we perform a 5-Fold Cross-Validation evaluation on a classification model (Logistic Regression / Random Forest) to obtain an unbiased, reliable performance estimate.

### Key Tasks:
1. **Step 1**: Train and evaluate a classification model using 5-Fold Cross-Validation (manual fold loop).
2. **Step 2**: Calculate and report the **mean** and **standard deviation** of the validation scores across all 5 folds ($Mean \pm Std$).
3. **Step 3**: Compare the Cross-Validated score against a single train/validation split score and analyze differences.
4. **Step 4**: Verify that **Stratified K-Fold** is used for classification and justify its importance in maintaining class proportion.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split, StratifiedKFold, KFold
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, f1_score

SEED = 42
np.random.seed(SEED)

# generate a synthetic classification dataset (Customer Churn simulation - week 4)
X, y = make_classification(
    n_samples=1000,
    n_features=10,
    n_informative=6,
    n_redundant=2,
    weights=[0.8, 0.2], # 80% Non-Churn, 20% Churn
    random_state=SEED
)

# convert to pandas df 
feature_names = [f"feature_{i+1}" for i in range(10)]
df = pd.DataFrame(X, columns=feature_names)
df['target'] = y

print(f"Dataset Shape: {df.shape}")
print(f"Class Distribution:\n{df['target'].value_counts(normalize=True)}")
df.head()
```

## Step 1 & Step 2: 5-Fold Cross-Validation (Manual Fold Loop)

Instead of relying on a single train/test split (which can vary depending on `random_state`), **K-Fold Cross-Validation** splits the dataset into $K$ equal subsets (folds). The model is trained on $K-1$ folds and evaluated on the remaining fold, repeating this process $K$ times.

Here, we evaluate our classifier using **5-Fold Cross-Validation** with the **weighted F1** metric and compute the mean and standard deviation:
$$\text{Score} = \text{Mean} \pm \text{Std}$$

```python
# initialize scaler and classifier separately (no Pipeline)
scaler = StandardScaler()
model = LogisticRegression(random_state=SEED)

# define 5-Fold Stratified Cross-Validation scheme
cv_stratified = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)

# 1: Manual 5-fold cross-validation — fit the scaler inside each fold to prevent data leakage
cv_scores = []
for train_idx, val_idx in cv_stratified.split(X, y):
    X_train_fold, X_val_fold = X[train_idx], X[val_idx]
    y_train_fold, y_val_fold = y[train_idx], y[val_idx]

    # fit scaler ONLY on the training fold, then transform both folds
    scaler.fit(X_train_fold)
    X_train_fold_scaled = scaler.transform(X_train_fold)
    X_val_fold_scaled = scaler.transform(X_val_fold)

    # train the model on the scaled training fold
    model.fit(X_train_fold_scaled, y_train_fold)
    y_val_pred = model.predict(X_val_fold_scaled)

    cv_scores.append(f1_score(y_val_fold, y_val_pred, average='weighted'))

cv_scores = np.array(cv_scores)

# 2: Report mean and std
mean_score = np.mean(cv_scores)
std_score = np.std(cv_scores)

print("---5-Fold Cross-Validation Results (Leak-Free Manual Scaling)---")
for fold_idx, score in enumerate(cv_scores, 1):
    print(f"Fold {fold_idx}: F1-Score = {score:.4f}")
print("-" * 40)
print(f"Mean F1-Score : {mean_score:.4f}")
print(f"Standard Dev  : ±{std_score:.4f}")
print(f"Final Estimate: {mean_score:.4f} ± {std_score:.4f}")
```

## Step 3: Comparing Cross-Validation Estimate vs. Single Train/Test Split Score

A single split evaluation evaluates performance on only one arbitrary subset of data. If the split happens to be "easy" or "difficult", the score will be overoptimistic or overly pessimistic. 

Below, we compute a single 80/20 train/validation split score and compare it with our 5-fold cross-validation mean score.

```python
# Perform a single 80/20 split on raw X
X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.20, random_state=SEED, stratify=y)

# Fit scaler on the train set only, then transform both sets (no Pipeline)
scaler = StandardScaler()
model = LogisticRegression(random_state=SEED)

scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_val_scaled = scaler.transform(X_val)

model.fit(X_train_scaled, y_train)

y_val_pred = model.predict(X_val_scaled)
single_split_f1 = f1_score(y_val, y_val_pred, average='weighted')

print(f"Single-Split F1-Score (Day 1 approach): {single_split_f1:.4f}")
print(f"5-Fold CV Mean F1-Score (Day 2 approach) : {mean_score:.4f} (±{std_score:.4f})")
print(f"Difference (Single Split - CV Mean)      : {single_split_f1 - mean_score:+.4f}")

# Visualizing Fold Scores vs Single Split
plt.figure(figsize=(8, 4))
plt.plot(range(1, 6), cv_scores, marker='o', linewidth=2, label='CV Fold Scores')
plt.axhline(mean_score, color='green', linestyle='--', label=f'CV Mean ({mean_score:.4f})')
plt.axhline(single_split_f1, color='red', linestyle=':', label=f'Single Split ({single_split_f1:.4f})')
plt.title('5-Fold Cross-Validation vs Single Split F1-Score')
plt.xlabel('Fold Number')
plt.ylabel('F1-Score')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## Step 3 & Step 4: Analytical Insights & Justification

### 1 Comparison of Estimates (Step 3 Explanation)
* **Single-Split Score**: Evaluates model performance on a single static test subset. In our run, the single-split F1-score was **0.8227**, which is lower than the cross-validation mean (-0.0275 difference). This happens because the specific 20% validation split was slightly more challenging than average.
* **Cross-Validation Estimate**: Provides a significantly more robust, realistic, and trustworthy measure of generalization (**0.8502**) because **every single data sample** is used in the validation set exactly once across the 5 iterations.
* **Interpretation of Standard Deviation ($\sigma$)**: The reported standard deviation ($\pm 0.0176$) quantifies the variance of performance across different data subsets. A small standard deviation confirms model stability across all 5 folds.

---

### 2.Confirmation & Importance of Stratified Folds (Step 4)
* **Confirmation**: `scikit-learn`'s default cross-validation strategy for classification estimators is `StratifiedKFold` (used automatically when passing an integer like `cv=5`). In our code, we explicitly defined `StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)`.
* **Why Stratification Matters**:
  1. **Class Ratio Preservation**: Our dataset is imbalanced (80% Class 0 vs 20% Class 1). Regular $K$-Fold creates folds randomly, which could lead to a fold having 5% of minority samples and another having 35%.
  2. **Elimination of Evaluation Bias**: `StratifiedKFold` guarantees that **every fold retains the exact 80/20 class proportions** present in the complete dataset.
  3. **Prevention of Training Failures**: Without stratification, a fold might completely miss instances of the minority class in its validation set or training split, making the F1-score computation invalid or extremely noisy.
